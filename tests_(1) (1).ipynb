# **Task 2**

```python
import task_2 as t2
import matplotlib.pyplot as plt
import numpy as np

xray = t2.load_image('images/task2_image.png')
plt.figure(figsize=(5, 5))
plt.imshow(xray, cmap='gray')
plt.title('Original X-Ray')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Run all three pipelines and store results
p1 = t2.pipeline_unsharp_masking(xray, blur_size=5, sigma=1.0, strength=1.5)
p2 = t2.pipeline_clahe_laplacian(xray, clip_limit=2.0, tile_grid_size=(8, 8), strength=1.0)
p3 = t2.pipeline_fft_highpass(xray, cutoff=30, boost=2.0)
```

```python
fig, axes = plt.subplots(1, 4, figsize=(20, 5))
images = [xray, p1, p2, p3]
titles = [
    'Original',
    'P1: Unsharp Masking',
    'P2: CLAHE + Laplacian',
    'P3: FFT High-Pass'
]
for ax, img, title in zip(axes, images, titles):
    ax.imshow(img, cmap='gray')
    ax.set_title(title, fontsize=11)
    ax.axis('off')
plt.suptitle('X-Ray Enhancement , Pipeline Comparison', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

# **Task 5**


```python
import task_5 as t5
import cv2
import matplotlib.pyplot as plt
```

```python
img1_path = "/content/London panorama (1).jpeg"
img2_path = "/content/London panorama.jpeg"
```

```python
img1, img2, img1_rgb, img2_rgb, gray1, gray2 = t5.load_images(img1_path, img2_path)

kp1, kp2, des1, des2 = t5.detect_features(gray1, gray2)

good_matches = t5.match_features(des1, des2)

H, mask = t5.compute_homography(kp1, kp2, good_matches)

result = t5.stitch_images(img1, img2, H)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img1_rgb)
plt.title("Image 1")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(img2_rgb)
plt.title("Image 2")
plt.axis("off")

plt.show()
```

```python
img1_kp = cv2.drawKeypoints(img1_rgb, kp1, None, color=(0,255,0))
img2_kp = cv2.drawKeypoints(img2_rgb, kp2, None, color=(0,255,0))

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img1_kp)
plt.title("Keypoints 1")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(img2_kp)
plt.title("Keypoints 2")
plt.axis("off")

plt.show()


matched_img = cv2.drawMatches(
    img1_rgb, kp1,
    img2_rgb, kp2,
    good_matches[:50], None,
    flags=2
)

plt.figure(figsize=(12,6))
plt.imshow(matched_img)
plt.title("Feature Matching")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
plt.title("Final Panorama")
plt.axis("off")
plt.show()
```

# **Task 8**

```python
import cv2
import matplotlib.pyplot as plt
import task_8 as t8
import numpy as np
```

```python
files = ["images/img_0.033.jpg", "images/img_0.25.jpg", "images/img_2.5.jpg", "images/img_15.jpg"]
times = np.array([1/30.0, 0.25, 2.5, 15.0], dtype=np.float32)

# ── 1. Show all input exposures ──────────────────────────────────────────────
imgs = [cv2.imread(f) for f in files]

fig, axes = plt.subplots(1, len(files), figsize=(20, 4))
fig.suptitle("Input Exposures", fontsize=14, fontweight='bold')
for ax, img, t in zip(axes, imgs, times):
    ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(f"{t:.4f}s")
    ax.axis('off')
plt.tight_layout()
plt.show()

# ── 2. custom pipeline ───────────────────────────────────────────────────
my_result = t8.run_hdr_pipeline(files, times)

# ── 3. OpenCV built-in pipeline ──────────────────────────────────────────────
# Align
align = cv2.createAlignMTB()
align.process(imgs, imgs)

# Calibrate CRF
calibrate = cv2.createCalibrateDebevec()
response = calibrate.process(imgs, times)

# Merge
merge = cv2.createMergeDebevec()
hdr_cv = merge.process(imgs, times, response)

# Tone map — using Reinhard to match your approach
tonemap = cv2.createTonemapReinhard(gamma=2.2, intensity=0.0, light_adapt=0.0, color_adapt=0.0)
cv2.result = tonemap.process(hdr_cv)
opencv_result = np.uint8(np.clip(cv2.result * 255, 0, 255))

# ── 4. Side-by-side comparison ───────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(16, 7))
fig.suptitle("HDR Comparison", fontsize=16, fontweight='bold')

axes[0].imshow(cv2.cvtColor(my_result, cv2.COLOR_BGR2RGB))
axes[0].set_title("Custom Pipeline", fontsize=13)
axes[0].axis('off')

axes[1].imshow(cv2.cvtColor(opencv_result, cv2.COLOR_BGR2RGB))
axes[1].set_title("OpenCV Built-in (Debevec + Reinhard)", fontsize=13)
axes[1].axis('off')

plt.tight_layout()
plt.show()
```
